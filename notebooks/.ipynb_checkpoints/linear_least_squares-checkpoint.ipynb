```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import json, pickle, pathlib, warnings
warnings.filterwarnings("ignore")
```

```python
ROOT = pathlib.Path("..").resolve()
DATA = ROOT / "data" 
PRED_DIR = ROOT / "data" / "predictions"
FIG_DIR  = ROOT / "figures" / "day05"
FIG_DIR.mkdir(parents=True, exist_ok=True)
```

```python
with open(ROOT / "data" / "model_config.json") as f:
    cfg = json.load(f)
tar_cfg = cfg["targets"]["tar_02084000"]

with open(ROOT / "data" / "norm_params.pkl", "rb") as f:
    nparams = pickle.load(f)["tar_02084000"]
```

```python
HORIZONS = tar_cfg['horizons']
FLOOD_THRESHOLD = cfg['flood_thresholds']['02084000']
FLOOD_WEIGHT = 10.0 

print('Days for Forecasting: ', f'{HORIZONS}')
print('Flood Threshold in Cubic Flow per Second: ', f'{FLOOD_THRESHOLD}')
```

```python
mean_lf = nparams["mean"]["log_flow"]
std_lf  = nparams["std"]["log_flow"]
```

```python
# local columns
local_cols = ["log_flow_norm", "prcp_mm_norm", "prcp_7day_norm", "prcp_14day_norm"]

# load splits 
splits = {}
for name in ['train', 'val', 'test']:
    splits[name] = pd.read_parquet(DATA / f"{name}.parquet")
    
# computing the normalized targets 
for name, df in splits.items():
    for h in HORIZONS: 
        df[f"target_norm_t{h}"] = ((df[f"target_log_flow_t{h}"] - mean_lf) / std_lf).astype("float32")

# upstream colums 
up_flow_cols = sorted([c for c in splits["train"].columns if c.startswith("up_flow_") and c.endswith("_norm")])
basin_cols   = sorted([c for c in splits["train"].columns if c.startswith("basin_prcp_") and c.endswith("_norm")])
upstream_cols = local_cols + up_flow_cols + basin_cols

print(f"Local features:    {len(local_cols)}   → A_local  will be n × {len(local_cols)+1} (with bias)")
print(f"Upstream features: {len(upstream_cols)} → A_upstream will be n × {len(upstream_cols)+1} (with bias)")
```

```python
def build_matrix(df, feature_cols, target_col):
    mask = df[target_col].notna()
    valid = df[mask]

    X = valid[feature_cols].values.astype(np.float64)
    b = valid[target_col].values.astype(np.float64)

    X = np.nan_to_num(X, nan=0.0)

    # Add bias column
    n = X.shape[0]
    A = np.column_stack([np.ones(n), X])

    return A, b, valid.index

matrices = {} 
model_features = {
    "local": local_cols,
    "upstream": upstream_cols,
}

for model_type, feat_cols in model_features.items():
    for h in HORIZONS:
        target_col = f"target_norm_t{h}"
        A, b, idx = build_matrix(splits["train"], feat_cols, target_col)
        matrices[(model_type, h)] = {"A": A, "b": b, "idx": idx}

print(f"  {'Model':<12} {'Horizon':<10} {'A shape':<18} {'b shape':<12} {'Rank check'}")
print(f"  {'─'*12} {'─'*10} {'─'*18} {'─'*12} {'─'*12}")
for (mt, h), m in sorted(matrices.items()):
    A = m["A"]
    rank = np.linalg.matrix_rank(A)
    print(f"  {mt:<12} t+{h:<8} {str(A.shape):<18} {str(m['b'].shape):<12} rank={rank}/{A.shape[1]}")
```

```python
# Solving with 2 equations 
solutions_ols = {}
for (model_type, h), m in sorted(matrices.items()):
    A, b = m["A"], m["b"]

    # Normal equations  →  (A^T A) x = A^T b
    x_normal = np.linalg.solve((A.T @ A), (A.T @ b))
    
    # QR Decomp of A 
    Q, R = np.linalg.qr(A)
    x_qr = np.linalg.solve(R, Q.T @ b)

    agree_nq = np.allclose(x_normal, x_qr, atol=1e-10)
    solutions_ols[(model_type, h)] = x_qr

    print(f"{model_type:>10} t+{h}: Normal≈QR: {agree_nq} "
          f"‖residual‖² = {np.sum((A @ x_qr - b)**2):.4f}")
```

```python
# weighted least squares
# Standard OLS weights everyday equally 
# Weighted LS: multiply each row by sqrt(w_i), then solve the transformed system
solutions_wls = {}   # (model_type, horizon) → x_wls

for (model_type, h), m in sorted(matrices.items()):
    A, b, idx = m["A"], m["b"], m["idx"]
    log_flow_target = b * std_lf + mean_lf # unnormalize
    flow_cfs = np.expm1(log_flow_target)
    is_flood = (flow_cfs > FLOOD_THRESHOLD).astype(float)

    w = np.where(is_flood > 0.5, np.sqrt(FLOOD_WEIGHT), 1.0)
    
    #   A_w = W^{1/2} A,  b_w = W^{1/2} b
    A_w = A * w[:, np.newaxis]
    b_w = b * w

    x_wls, _, _, _ = np.linalg.lstsq(A_w, b_w, rcond=None)
    solutions_wls[(model_type, h)] = x_wls

    n_flood = int(is_flood.sum())
    x_ols = solutions_ols[(model_type, h)]
    max_diff = np.max(np.abs(x_ols - x_wls))

    print(f"{model_type:>10} t+{h}:  flood_days={n_flood:>4}/{len(b)}  "
          f"max|x_ols - x_wls| = {max_diff:.4f}")
    
```

```python
# ŷ_norm = A @ x_wls
def predict(split_name, df, model_type, horizon, x_wls, feat_cols):
    target_col = f"target_norm_t{horizon}"
    target_raw_col = f"target_log_flow_t{horizon}"

    mask = df[target_col].notna()
    valid = df[mask]

    X = valid[feat_cols].values.astype(np.float64)
    X = np.nan_to_num(X, nan=0.0)
    A = np.column_stack([np.ones(len(X)), X])

    y_pred_norm = A @ x_wls
    log_predicted = y_pred_norm * std_lf + mean_lf
    predicted_cfs = np.expm1(log_predicted)
    predicted_cfs = np.maximum(predicted_cfs, 0.0) # non negatives

    log_observed = valid[target_raw_col].values
    observed_cfs = np.expm1(log_observed)

    is_flood = (observed_cfs > FLOOD_THRESHOLD).astype(int)
    
    # Network state: count upstream gauges above their flood thresholds
    up_gauges = tar_cfg["upstream_gauges"]
    net_state = np.zeros(len(valid), dtype=int)
    for ug in up_gauges:
        col = f"up_flow_{ug}"
        if col in valid.columns:
            ug_thresh = cfg["flood_thresholds"].get(ug, 1e9)
            ug_flow = np.expm1(valid[col].values)
            net_state += (ug_flow > ug_thresh).astype(int)
    
    # Build output dataframe
    out = pd.DataFrame({
        "date": valid.index,
        "observed_cfs": observed_cfs,
        "predicted_cfs": predicted_cfs,
        "log_observed": log_observed,
        "log_predicted": log_predicted,
        "is_flood": is_flood,
        "network_state": net_state,
        "horizon": horizon,
    })
    
    return out

predictions = {}  

for model_type, feat_cols in model_features.items():
    for h in HORIZONS:
        x_wls = solutions_wls[(model_type, h)]
        for split_name in ["val", "test"]:
            df = splits[split_name]
            out = predict(split_name, df, model_type, h, x_wls, feat_cols)
            predictions[(model_type, h, split_name)] = out
    
```

```python
fig, axes = plt.subplots(2, 4, figsize=(18, 8), sharey="row")

residual_stats = {}

for i, model_type in enumerate(["local", "upstream"]):
    for j, h in enumerate(HORIZONS):
        ax = axes[i, j]
        
        # Val residuals (used for calibration)
        df_val = predictions[(model_type, h, "val")]
        r = df_val["log_observed"].values - df_val["log_predicted"].values
        r = r[np.isfinite(r)]
        
        residual_stats[(model_type, h)] = r
        
        # Histogram
        ax.hist(r, bins=50, color="steelblue", alpha=0.7, edgecolor="white", density=True)
        
        # Quantiles for interval
        q_lo = np.quantile(r, 0.05)
        q_hi = np.quantile(r, 0.95)
        ax.axvline(q_lo, color="red", ls="--", lw=1.5, label=f"q5 = {q_lo:.3f}")
        ax.axvline(q_hi, color="red", ls="--", lw=1.5, label=f"q95 = {q_hi:.3f}")
        ax.axvline(0, color="black", ls="-", lw=0.8, alpha=0.5)
        
        ax.set_title(f"{model_type.title()} OLS — t+{h}")
        ax.legend(fontsize=7)
        if j == 0:
            ax.set_ylabel("Density")
        if i == 1:
            ax.set_xlabel("Signed residual (log scale)")

plt.suptitle("Validation Set Residuals — Linear Models", fontsize=13, y=1.02)
plt.tight_layout()
plt.savefig(FIG_DIR / "residual_histograms.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
print(f"\n{'Model':<15} {'Horizon':<8} {'Mean':>8} {'Std':>8} {'Min':>8} {'Max':>8} {'q5':>8} {'q95':>8}")
print(f"{'─'*15} {'─'*8} {'─'*8} {'─'*8} {'─'*8} {'─'*8} {'─'*8} {'─'*8}")
for (mt, h), r in sorted(residual_stats.items()):
    print(f"{mt+' OLS':<15} t+{h:<6} {np.mean(r):>8.3f} {np.std(r):>8.3f} "
          f"{np.min(r):>8.3f} {np.max(r):>8.3f} {np.quantile(r,0.05):>8.3f} {np.quantile(r,0.95):>8.3f}")
```

```python
# building prediction intervals from residuals 

Alpha = 0.10   # α = 0.10 → 90% prediction interval
interval_results = [] 

for model_type in ['local', 'upstream']:
    for h in HORIZONS: 
        r = residual_stats[(model_type, h)]
        n = len(r)

        level_lo = max(np.floor((n + 1) * (Alpha / 2)) / n, 0.0)
        level_hi = min(np.ceil((n + 1) * (1 - Alpha / 2)) / n, 1.0)
        q_lo = np.quantile(r, level_lo)
        q_hi = np.quantile(r, level_hi)

        df_test = predictions[(model_type, h, "test")]
        log_pred = df_test["log_predicted"].values
        log_obs  = df_test["log_observed"].values
        obs_cfs  = df_test["observed_cfs"].values
        is_flood = df_test["is_flood"].values > 0.5

        lower = np.expm1(log_pred + q_lo)
        upper = np.expm1(log_pred + q_hi)
        lower = np.maximum(lower, 0.0)

        # prediction coverage
        covered = (obs_cfs >= lower) & (obs_cfs <= upper)
        coverage_all   = covered.mean()
        coverage_flood = covered[is_flood].mean() if is_flood.sum() > 0 else np.nan

        # prediction interval width
        width = upper - lower
        width_all   = np.median(width)
        width_flood = np.median(width[is_flood]) if is_flood.sum() > 0 else np.nan

        interval_results.append({
            "Model": f"{model_type.title()} OLS",
            "Horizon": f"t+{h}",
            "q_lo": q_lo,
            "q_hi": q_hi,
            "Coverage_all": coverage_all,
            "Coverage_flood": coverage_flood,
            "Width_all": width_all,
            "Width_flood": width_flood,
        })

interval_df = pd.DataFrame(interval_results)

print("═" * 80)
print("Prediction Intervals — Linear Models (Test Set, 90% target)")
print("═" * 80)
print(interval_df.to_string(index=False, float_format="%.3f"))
```

```python
# NSE  = 1 − Σ(obs−pred)² / Σ(obs−mean_obs)²  
# RMSE = √mean((obs−pred)²)  in cfs

def nse(obs, pred):
    num = np.sum((obs - pred) ** 2)
    den = np.sum((obs - np.mean(obs)) ** 2)
    return 1.0 - num / den if den > 0 else np.nan

def rmse(obs, pred):
    return np.sqrt(np.mean((obs - pred) ** 2))

# Persistence baseline: predict today's observed flow as the t+k forecast
perf_rows = []

for split_name in ["val", "test"]:
    df_base = splits[split_name]

    for h in HORIZONS:
        target_raw = f"target_log_flow_t{h}"
        mask = df_base[target_raw].notna()
        obs_cfs  = np.expm1(df_base.loc[mask, target_raw].values)
        pers_cfs = np.expm1(df_base.loc[mask, "log_flow"].values)
        is_flood = obs_cfs > FLOOD_THRESHOLD

        perf_rows.append({
            "split": split_name, "model": "Persistence", "horizon": h,
            "nse_all":   nse(obs_cfs, pers_cfs),
            "rmse_all":  rmse(obs_cfs, pers_cfs),
            "nse_flood":  nse(obs_cfs[is_flood], pers_cfs[is_flood]),
            "rmse_flood": rmse(obs_cfs[is_flood], pers_cfs[is_flood]),
        })

    for model_type in ["local", "upstream"]:
        for h in HORIZONS:
            df_pred = predictions[(model_type, h, split_name)]
            obs  = df_pred["observed_cfs"].values
            pred = df_pred["predicted_cfs"].values
            is_flood = df_pred["is_flood"].values > 0.5

            perf_rows.append({
                "split": split_name,
                "model": f"{model_type.title()} OLS",
                "horizon": h,
                "nse_all":    nse(obs, pred),
                "rmse_all":   rmse(obs, pred),
                "nse_flood":  nse(obs[is_flood], pred[is_flood]),
                "rmse_flood": rmse(obs[is_flood], pred[is_flood]),
            })

perf_df = pd.DataFrame(perf_rows)

for split_name in ["test"]:
    sub = perf_df[perf_df["split"] == split_name]
    print(f"\n{'═'*72}")
    print(f"  Forecast Performance — {split_name.upper()} set")
    print(f"{'═'*72}")

    # NSE — all days
    print(f"\n  NSE (all days):")
    print(f"  {'Model':<18} {'t+1':>7} {'t+2':>7} {'t+3':>7} {'t+4':>7}")
    print(f"  {'─'*18} {'─'*7} {'─'*7} {'─'*7} {'─'*7}")
    for model in ["Persistence", "Local OLS", "Upstream OLS"]:
        row = sub[sub["model"] == model].set_index("horizon")["nse_all"]
        vals = "  ".join(f"{row.get(h, float('nan')):>7.3f}" for h in HORIZONS)
        print(f"  {model:<18} {vals}")

    # NSE — flood days only
    print(f"\n  NSE (flood days only):")
    print(f"  {'Model':<18} {'t+1':>7} {'t+2':>7} {'t+3':>7} {'t+4':>7}")
    print(f"  {'─'*18} {'─'*7} {'─'*7} {'─'*7} {'─'*7}")
    for model in ["Persistence", "Local OLS", "Upstream OLS"]:
        row = sub[sub["model"] == model].set_index("horizon")["nse_flood"]
        vals = "  ".join(f"{row.get(h, float('nan')):>7.3f}" for h in HORIZONS)
        print(f"  {model:<18} {vals}")

    # RMSE — all days
    print(f"\n  RMSE in cfs (all days):")
    print(f"  {'Model':<18} {'t+1':>9} {'t+2':>9} {'t+3':>9} {'t+4':>9}")
    print(f"  {'─'*18} {'─'*9} {'─'*9} {'─'*9} {'─'*9}")
    for model in ["Persistence", "Local OLS", "Upstream OLS"]:
        row = sub[sub["model"] == model].set_index("horizon")["rmse_all"]
        vals = "  ".join(f"{row.get(h, float('nan')):>9.1f}" for h in HORIZONS)
        print(f"  {model:<18} {vals}")
```

```python
# Weight interpretation via QR 

from matplotlib.patches import Patch

def feature_labels(model_type):
    return ["bias"] + model_features[model_type]

def feature_group(col_name):
    if col_name == "bias":                  return "bias"
    if col_name.startswith("up_flow_"):     return "upstream"
    if col_name.startswith("basin_prcp_"):  return "basin precip"
    return "local"

group_colors = {
    "local":        "#1f77b4",
    "upstream":     "#2ca02c",
    "basin precip": "#ff7f0e",
}

model_type = "upstream"
labels  = feature_labels(model_type)
colors  = [group_colors.get(feature_group(c), "#888888") for c in labels]

# Vertical layout: 4 rows, 1 column
fig, axes = plt.subplots(4, 1, figsize=(16, 14), sharey=True)
contrib_summary = {}

for j, h in enumerate(HORIZONS):
    ax = axes[j]

    A = matrices[(model_type, h)]["A"]
    b = matrices[(model_type, h)]["b"]

    # QR and contribution vector
    Q, R = np.linalg.qr(A)
    c = Q.T @ b
    share = (c ** 2) / np.sum(c ** 2)

    # Drop bias from the plot
    share_features  = share[1:] / share[1:].sum()
    labels_features = labels[1:]
    colors_features = colors[1:]

    contrib_summary[h] = {"labels": labels, "c": c, "share": share}

    x_pos = np.arange(len(labels_features))
    ax.bar(x_pos, share_features, color=colors_features, edgecolor="white")

    ax.set_xticks(x_pos)
    ax.set_xticklabels(labels_features, rotation=75, fontsize=7, ha="right")
    ax.set_title(f"Upstream model — t+{h}", fontsize=11)
    ax.set_ylabel("Variance share\n(excl. bias)")

handles = [Patch(facecolor=v, label=k) for k, v in group_colors.items()]
fig.legend(
    handles=handles,
    loc="upper center",
    ncol=3,
    fontsize=9,
    bbox_to_anchor=(0.5, 1.01)
)

plt.suptitle(
    r"Upstream model — QR contributions $c_i = q_i^\top b$",
    fontsize=13,
    y=1.035
)

plt.tight_layout()
plt.savefig(FIG_DIR / "qr_contributions_upstream_vertical.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Top-5 non-bias contributors per horizon
print("\nTop-5 contributors (excl. bias) by variance share — UPSTREAM:")
print("─" * 70)

for h in HORIZONS:
    info = contrib_summary[h]
    share = info["share"][1:]
    labels_h = info["labels"][1:]

    share_renorm = share / share.sum()
    order = np.argsort(share_renorm)[::-1][:5]

    items = "  ".join(
        f"{labels_h[k]}={share_renorm[k]*100:.1f}%"
        for k in order
    )

    print(f"  t+{h}: {items}")
```
